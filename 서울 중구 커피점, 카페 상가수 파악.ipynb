```python
query = ''' 
select sido_nm,
    sgg_nm,
    upjong_mnm,
   opn_1 + opn_2 + opn_3 + opn_5 + opn_5up as total  -- 총 상가수를 더해서 total로 저장하기
from sgg_upjong_cnt 
where sido_nm = "서울특별시"
and sgg_nm = "중구"
and upjong_mnm = "커피점/카페"
'''
```
