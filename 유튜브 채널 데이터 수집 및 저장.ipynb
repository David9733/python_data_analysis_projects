```python
from bs4 import BeautifulSoup 
from selenium import webdriver 
import pandas as pd 
import time 
```

```python
browser = webdriver.Chrome()
time.sleep(1)
url = 'https://youtube-rank.com/board/bbs/board.php?bo_table=youtube'
browser.get(url)
```

```python
html = browser.page_source
soup = BeautifulSoup(html, 'html.parser')
```

```python
lst = soup.select('tr.aos-init')[0:3] 
```

```python
result = []

for element in lst: 
    ranking = element.select('td.rank')[0].text.strip() 
    category = element.select('p.category')[0].text.strip()
    youtube_name = element.select('h1 > a')[0].text.strip()
    subscriber_count = element.select('td.subscriber_cnt')[0].text
    view_count = element.select('td.view_cnt')[0].text
    video_count = element.select('td.video_cnt')[0].text

    element_lst = [ranking, category, youtube_name, subscriber_count, view_count, video_count]
    
    result.append(element_lst)
```

```python
print(result)
```
