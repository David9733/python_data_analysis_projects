```python
import pandas as pd 
from sqlalchemy import create_engine 
```

```python
# MariaDB 연결 설정
username = '***'    # MariaDB 사용자 이름
password = '***'   # MariaDB 비밀번호
host = '***'   # MariaDB 호스트
port = '***'            # MariaDB 포트
database = '***'         # 접속할 데이터베이스 이름

# SQLAlchemy 엔진 생성
engine = create_engine(f'mysql+pymysql://{username}:{password}@{host}:{port}/{database}')
```

```python
query = '''
select * 
from customers
where join_date = (select max(join_date)  
                        from customers)
'''
```

```python
df = pd.read_sql_query(query, engine)
print(df)  # 현재 접속중인 페이지 URL 확인하는 코드
```
