```python
# 함수로 만든 버전
```

```python
def yt_review(url, content_name, end_key_input): 
    # 1. url 로 이동하기 
    browser.get(url)
    # 3초 대기하기 
    time.sleep(3) 
    
    # 2. end 키 누르기 
    for i in range(end_key_input): 
        # end 키 한번 입력
        browser.find_element('css selector', 'body').send_keys(Keys.END)
        # 1초 대기하기 
        time.sleep(1)
    
    # 3. html 수집 (snap-shot) 
    html = browser.page_source 
    # 4. BeautifulSoup 형태로 변환 
    soup = BeautifulSoup(html, 'html.parser')
    
    # 5. 정보뭉치 찾기 
    lst = soup.select('ytd-comment-thread-renderer.style-scope.ytd-item-section-renderer')
    
    # 7. 반복문으로 정보뭉치에서 정보 찾고, 저장하기 
    # 정보를 모을 공간 
    result = []

    # 반복문을 통해 80개의 댓글 저장하기 
    for element in lst: 
        reply = element.select('div#content')[0].text.strip()
        temp = [content_name, reply]
        result.append(temp)
    
    # 8.데이터 프레임으로 만들기 
    result_df = pd.DataFrame(result, columns=['컨텐츠명', '댓글'])
    
    # 출력하기 
    return result_df 
```

```python
df = yt_review('https://www.youtube.com/watch?v=kH86Df2bpyg', '인사이드 아웃', 4)
df
```
